## 1.1 — The 5 Core Data Types

```python
x = 42           # int — whole numbers
y = 3.14         # float — decimal numbers
name = "Akash"   # str — text, always in quotes
is_ready = True  # bool — True or False (capital T and F)
result = None    # None — the absence of a value

print(type(x))
print(type(y))
print(type(name))
print(type(is_ready))
print(type(result))
```

## 1.2 — Type Conversion

```python
# Converting between types explicitly
print(int("42"))       # str -> int
print(float("3.14"))   # str -> float
print(str(100))        # int -> str

# What happens when conversion fails
# int("hello")   -> ValueError: type is correct but value makes no sense
# int("3.14")    -> ValueError: not a whole number
# "5" + 3        -> TypeError: incompatible types for + operation
```

```python
# Exercise: add a string number and an integer, then join them as text
num_str = "7"
num_int = 4
print(int(num_str) + num_int)  # 11
print(num_str + str(num_int))  # "74"
```

## 1.3 — Variables and Naming

```python
# Good naming — snake_case, descriptive
user_name = "Akash"
total_score = 0
is_logged_in = False

# = is assignment (stores a value)
score = 5
score = score + 3
print(score)       # 8

# == is comparison (returns True or False)
print(score == 10) # False
print(score == 8)  # True
```

## 1.4 — Strings: Indexing, Slicing, F-strings

```python
name = "Akash"

# Indexing — starts at 0
print(name[0])    # "A"
print(name[1])    # "k"
print(name[-1])   # "h" — negative counts from end

# Slicing — [start:stop] stop is not included
print(name[0:3])  # "Aka"
print(name[2:])   # "ash"
print(name[:3])   # "Aka"

# F-strings
score = 95.678
print(f"Name: {name}")
print(f"Score: {score:.2f}")  # rounds to 2 decimal places
```

```python
# Exercise: name slicing with f-string
name = "Akash Chaudhari"
print(name[:5])                                          # "Akash"
print(name[-1])                                          # "i"
print(f"Name: {name}, Length: {len(name)}")              # dynamic length
```

## 1.5 — Lists

```python
scores = [95, 87, 72, 100]

# Accessing
print(scores[0])   # 95
print(scores[-1])  # 100

# Core methods
scores.append(88)         # adds to end
scores.pop()              # removes and returns last item (88 removed, discarded)
print(len(scores))        # 4
print(95 in scores)       # True
```

```python
# Exercise: cities list
cities = ["Mumbai", "Delhi", "Bangalore"]
cities.append("Chennai")
cities.pop()
print(cities)             # ['Mumbai', 'Delhi', 'Bangalore']
print(len(cities))        # 3
print("Delhi" in cities)  # True
```

## 1.6 — Dictionaries

```python
user = {
    "name": "Akash",
    "age": 28,
    "is_active": True
}

# Accessing
print(user["name"])   # "Akash"

# Modifying and adding
user["age"] = 29
user["city"] = "Mumbai"

# Safe access with .get()
# user["phone"]               -> KeyError: crashes
print(user.get("phone"))       # None — safe, no crash
print(user.get("phone", "N/A")) # "N/A" — custom default

# Dictionary methods
print(user.keys())
print(user.values())
print(user.items())
```

```python
# Exercise: job posting dict
job = {
    "title": "Software Engineer",
    "company": "TechCorp",
    "location": "New York",
    "remote": True
}

print(job["company"])
print(job.get("salary", "Not disclosed"))
job["experience_years"] = 3
print(job.keys())
```

## 1.7 — String Methods

```python
s = "  Hello, World!  "

print(s.strip())                   # removes whitespace from both ends
print(s.lower())                   # lowercase
print(s.upper())                   # uppercase
print(s.replace("World", "Akash")) # replace substring
print(s.find("World"))             # index where substring starts, -1 if not found

# split and join — learn as a pair
sentence = "Mumbai,Delhi,Bangalore"
cities = sentence.split(",")         # str -> list
result = ",".join(cities)            # list -> str
print(cities)
print(result)

# startswith / endswith
url = "https://google.com"
print(url.startswith("https"))   # True
print(url.endswith(".com"))      # True
```

```python
# Exercise: strip, split, join pipeline
s = "   llm,rag,langchain,langgraph   "
s = s.strip()
ai = s.split(",")
print(ai)
s = " | ".join(ai)
print(s)
```

## 1.8 — Combining Lists and Dicts

```python
# List of dicts — the pattern you'll see in every API response
agents = [
    {"name": "Planner", "active": True},
    {"name": "Researcher", "active": False},
    {"name": "Critic", "active": True}
]

print(agents[0])             # first dict
print(agents[0]["name"])     # "Planner"
print(agents[1]["active"])   # False
```

## Day 1 — Key Takeaways

- 5 core types: `int`, `float`, `str`, `bool`, `None`
- `=` assigns, `==` compares
- `TypeError` = incompatible types for an operation
- `ValueError` = type is right, value makes no sense
- `dict["key"]` crashes on missing key → use `dict.get("key", default)` instead
- `append()` adds to list end, `pop()` removes and returns last item
- `split()` turns a string into a list, `join()` turns a list into a string
- List of dicts is the pattern behind every API response you'll ever work with
