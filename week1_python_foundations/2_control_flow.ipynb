## 2.1 — Booleans and Comparison Operators

```python
# Comparison operators — always return a bool
print(5 == 5)   # True
print(5 != 3)   # True  — not equal
print(5 > 3)    # True
print(5 < 3)    # False
print(5 >= 5)   # True  — greater than or equal
print(5 <= 4)   # False

# Logical operators
print(True and False)  # False — both must be True
print(True or False)   # True  — at least one must be True
print(not True)        # False — flips the bool

# Real usage
age = 25
has_id = True
print(age >= 18 and has_id)   # True
print(age < 18 or has_id)     # True
print(not has_id)              # False
```

## 2.2 — Truthiness

```python
# Falsy values — treated as False in conditions
# False, None, 0, 0.0, "", [], {}

# Everything else is truthy

print(bool([]))       # False — empty list
print(bool(0))        # False — zero
print(bool(""))       # False — empty string
print(bool("Akash"))  # True
print(bool([1, 2]))   # True

# Pythonic usage — no need to write != ""
name = ""
if name:
    print("Name exists")
else:
    print("Name is empty")  # this runs
```

```python
# Exercise: check_profile using truthiness
def check_profile(profile):
    if not profile:
        print("Empty profile")
    elif profile.get("name"):
        print(f"Welcome, {profile['name']}!")
    else:
        print("Name is missing")

check_profile({})
check_profile({"name": "Akash"})
check_profile({"name": ""})
```

## 2.3 — if / elif / else

```python
# Python checks top to bottom, stops at first True
score = 72

if score >= 90:
    print("A grade")
elif score >= 80:
    print("B grade")
elif score >= 70:
    print("C grade")   # this runs
else:
    print("Below C")

# WARNING: order matters
# A broader condition placed before a narrower one will swallow it
# score >= 80 placed before score >= 85 means 85 never gets checked separately
```

```python
# Guard clauses — keep code flat instead of deeply nested

# Nested — harder to read
def process_nested(user):
    if user:
        if user.get("active"):
            print("Processing user")

# Guard clause — cleaner, same logic
def process_guard(user):
    if not user:
        return
    if not user.get("active"):
        return
    print("Processing user")
```

```python
# Exercise: classify_experience
def classify_experience(exp):
    if exp < 2:
        print("Junior")
    elif 2 <= exp <= 5:   # chained comparison — more idiomatic
        print("Mid-Level")
    else:
        print("Senior")

classify_experience(1)
classify_experience(3)
classify_experience(7)
```

## 2.4 — for Loops, range(), enumerate()

```python
# Basic for loop
skills = ["python", "sql", "docker"]
for skill in skills:
    print(skill)

# Looping over a dict
user = {"name": "Akash", "age": 28}
for key in user:
    print(key)                    # prints keys only

for key, value in user.items():
    print(key, value)             # prints key-value pairs

# range()
for i in range(3):
    print(i)        # 0, 1, 2

for i in range(2, 6):
    print(i)        # 2, 3, 4, 5

# enumerate() — index AND value together
for i, skill in enumerate(skills):
    print(i, skill)   # 0 python, 1 sql, 2 docker

# enumerate with start index
for i, skill in enumerate(skills, 1):
    print(i, skill)   # 1 python, 2 sql, 3 docker
```

```python
# Exercise: ranked list from list of dicts
result = [
    {"name": "Akash", "score": 95},
    {"name": "Ab", "score": 25},
    {"name": "As", "score": 9},
    {"name": "Ag", "score": 5}
]

for i, item in enumerate(result, 1):
    print(f"{i}. {item['name']} - {item['score']}")
```

## 2.5 — break and continue

```python
# break — exit the loop immediately
skills = ["python", "sql", "docker", "kubernetes"]
for skill in skills:
    if skill == "docker":
        break
    print(skill)
# prints: python, sql

# continue — skip current iteration, keep going
scores = [95, 0, 87, 0, 72]
for score in scores:
    if score == 0:
        continue
    print(score)
# prints: 95, 87, 72

# Rule:
# break  — found what you need, stop entirely
# continue — bad data, skip and keep processing
```

## 2.6 — while Loops

```python
# while runs as long as condition is True
count = 0
while count < 3:
    print(count)
    count += 1
# prints: 0, 1, 2

# Use for: iterating over a known collection
# Use while: waiting for a condition to change

# Exercise: password retry
passwords = ["wrong", "wrong", "correct"]
attempts = 0

while attempts < len(passwords):
    current_password = passwords[attempts]
    if current_password == "correct":
        print("Access granted")
        break
    else:
        print("Wrong password, trying again")
    attempts += 1
```

## Day 2 Build Target — Agent Monitor

```python
agents = [
    {"name": "Planner", "status": "active", "tasks_completed": 12},
    {"name": "Scout", "status": "inactive", "tasks_completed": 5},
    {"name": "Sentry", "status": "active", "tasks_completed": 8},
    {"name": "Ghost", "status": "active", "tasks_completed": 15},
    {"name": "Medic", "status": "inactive", "tasks_completed": 3}
]

total_active_tasks = 0

for i, agent in enumerate(agents, 1):
    if agent["status"] == "inactive":
        continue

    tasks = agent["tasks_completed"]
    print(f"Agent {i}: {agent['name']} — {tasks} tasks")

    total_active_tasks += tasks

    if tasks > 10:
        print(f"High performer: {agent['name']}")

print(f"\nTotal tasks completed by active agents: {total_active_tasks}")
```

## Day 2 — Key Takeaways

- Comparison operators (`==`, `!=`, `<`, `>`, `<=`, `>=`) always return a bool
- Falsy values: `False`, `None`, `0`, `0.0`, `""`, `[]`, `{}`
- `if not x:` is more Pythonic than `if x == []` or `if x == ""`
- `elif` order matters — broader conditions must come after narrower ones, or they swallow them
- Guard clauses keep code flat — return early instead of deep nesting
- `enumerate(list, 1)` gives index starting from 1 and the item together
- `break` exits the loop entirely, `continue` skips to the next iteration
- Use `for` over known collections, `while` when waiting for a condition to change
- Chained comparisons: `2 <= x <= 5` is valid and more readable than `x >= 2 and x <= 5`
- `== ` checks value equality, `is` checks memory identity (same object)
